```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML


def f(x, y, scale=1.0):

    scaled_x = x / scale
    scaled_y = y / scale
    return (scaled_x**2 + scaled_y**2 - 1)**3 - scaled_x**2 * scaled_y**3


x = np.linspace(-1.5, 1.5, 400)
y = np.linspace(-1.5, 1.5, 400)
X, Y = np.meshgrid(x, y)


fig, ax = plt.subplots(figsize=(8, 8))


ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect('equal', adjustable='box')


def update(frame):

    ax.clear()


    ax.set_title('Animating the Heart with text "Jesus Loves You"')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_aspect('equal', adjustable='box')
    ax.grid(True)
    ax.set_xlim(-1.5, 1.5)
    ax.set_ylim(-1.5, 1.5)


    scale_factor = 1.0 + 0.1 * np.sin(frame * 0.15)


    Z_new = f(X, Y, scale_factor)


    ax.contour(X, Y, Z_new, levels=[0], colors='red')


    ax.text(0, -0.2, 'Jesus Loves You',
            fontsize=20,
            color='darkred',
            ha='center',
            va='center',
            fontweight='bold',
            bbox=dict(boxstyle="round,pad=0.5", fc='lightpink', ec='darkred', lw=1, alpha=0.6))


    return []


ani = FuncAnimation(fig, update, frames=100, interval=100, blit=False)


html_output = HTML(ani.to_jshtml())
plt.close(fig)
html_output
```
