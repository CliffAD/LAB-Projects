```
# Module 7 – Business Leadership
Estimated duration: 2 minutes

### 0:00–0:12 — Opening
* **Visual:** A diverse team entering a modern workplace, collaborating around a table.
* **On-screen text:** Module 7 – Business Leadership
* **Narration:** "Business leadership is more than having a title. It is the ability to guide people, communicate with purpose, make sound decisions, solve challenges, and act with integrity."

### 0:12–0:32 — Leadership
* **Visual:** A team leader setting goals and encouraging team members.
* **On-screen text:** Leadership: Set direction. Inspire action.
* **Narration:** "Effective leaders create a clear direction and help others understand why the goal matters. They lead by example, take responsibility, recognize team strengths, and motivate people to perform at their best. Leadership is not about controlling every detail. It is about building trust and helping the team succeed together."

### 0:32–0:52 — Communication
* **Visual:** A manager listening to an employee, followed by a clear team briefing.
* **On-screen text:** Communication: Listen. Clarify. Connect.
* **Narration:** "Strong communication begins with active listening. Leaders explain ideas clearly, ask thoughtful questions, and adapt their message to different audiences. They create space for feedback and make sure everyone understands their role. Clear communication reduces confusion, strengthens relationships, and keeps teams aligned."

### 0:52–1:12 — Decision Making
* **Visual:** Team members reviewing information on a screen and choosing between options.
* **On-screen text:** Decision Making: Assess. Choose. Take responsibility.
* **Narration:** "Business leaders make decisions using reliable information, sound judgment, and an awareness of risk. They consider different perspectives, evaluate possible outcomes, and avoid delaying important choices unnecessarily. Once a decision is made, strong leaders communicate it clearly and remain accountable for the result."

### 1:12–1:32 — Problem Solving
* **Visual:** A team examining a workflow, identifying a bottleneck, and testing a solution.
* **On-screen text:** Problem Solving: Find the cause. Build the solution.
* **Narration:** "Good problem solving starts by defining the real issue. Instead of treating only the symptoms, leaders look for the root cause. They generate practical options, test solutions, learn from setbacks, and improve the process. A challenge can become an opportunity when a team responds with curiosity and discipline."

### 1:32–1:52 — Business Ethics
* **Visual:** An employee choosing transparency in a difficult workplace situation.
* **On-screen text:** Business Ethics: Do what is right.
* **Narration:** "Business ethics means acting with honesty, fairness, transparency, and respect. Ethical leaders make responsible choices even when shortcuts seem easier. They protect trust, treat people fairly, follow the rules, and consider the impact of their decisions on customers, employees, communities, and the wider business."

### 1:52–2:00 — Closing
* **Visual:** The team completes a project and looks ahead confidently.
* **On-screen text:** Lead with purpose.\nCommunicate with clarity.\nDecide with integrity.
* **Narration:** "Leadership brings these skills together. Lead with purpose, communicate with clarity, solve problems thoughtfully, and make decisions with integrity."
```

```
import os
from gtts import gTTS
from PIL import Image, ImageDraw

# Script data with timing, visual slide text, and narration text
script_segments = [
    {
        "id": "01_opening",
        "time": "0:00–0:12",
        "title": "Module 7 – Business Leadership",
        "text": "Module 7 – Business Leadership",
        "narration": "Business leadership is more than having a title. It is the ability to guide people, communicate with purpose, make sound decisions, solve challenges, and act with integrity."
    },
    {
        "id": "02_leadership",
        "time": "0:12–0:32",
        "title": "Leadership",
        "text": "Leadership:\nSet direction. Inspire action.",
        "narration": "Effective leaders create a clear direction and help others understand why the goal matters. They lead by example, take responsibility, recognize team strengths, and motivate people to perform at their best. Leadership is not about controlling every detail. It is about building trust and helping the team succeed together."
    },
    {
        "id": "03_communication",
        "time": "0:32–0:52",
        "title": "Communication",
        "text": "Communication:\nListen. Clarify. Connect.",
        "narration": "Strong communication begins with active listening. Leaders explain ideas clearly, ask thoughtful questions, and adapt their message to different audiences. They create space for feedback and make sure everyone understands their role. Clear communication reduces confusion, strengthens relationships, and keeps teams aligned."
    },
    {
        "id": "04_decision_making",
        "time": "0:52–1:12",
        "title": "Decision Making",
        "text": "Decision Making:\nAssess. Choose. Take responsibility.",
        "narration": "Business leaders make decisions using reliable information, sound judgment, and an awareness of risk. They consider different perspectives, evaluate possible outcomes, and avoid delaying important choices unnecessarily. Once a decision is made, strong leaders communicate it clearly and remain accountable for the result."
    },
    {
        "id": "05_problem_solving",
        "time": "1:12–1:32",
        "title": "Problem Solving",
        "text": "Problem Solving:\nFind the cause. Build the solution.",
        "narration": "Good problem solving starts by defining the real issue. Instead of treating only the symptoms, leaders look for the root cause. They generate practical options, test solutions, learn from setbacks, and improve the process. A challenge can become an opportunity when a team responds with curiosity and discipline."
    },
    {
        "id": "06_business_ethics",
        "time": "1:32–1:52",
        "title": "Business Ethics",
        "text": "Business Ethics:\nDo what is right.",
        "narration": "Business ethics means acting with honesty, fairness, transparency, and respect. Ethical leaders make responsible choices even when shortcuts seem easier. They protect trust, treat people fairly, follow the rules, and consider the impact of their decisions on customers, employees, communities, and the wider business."
    },
    {
        "id": "07_closing",
        "time": "1:52–2:00",
        "title": "Closing",
        "text": "Lead with purpose.\nCommunicate with clarity.\nDecide with integrity.",
        "narration": "Leadership brings these skills together. Lead with purpose, communicate with clarity, solve problems thoughtfully, and make decisions with integrity."
    }
]

print("Generating audio narration files...")
for seg in script_segments:
    tts = gTTS(text=seg["narration"], lang='en')
    tts.save(f"narration/{seg['id']}.mp3")
    print(f" Saved narration: narration/{seg['id']}.mp3 ({seg['time']})")

print("\nGenerating slide template images (16:9 - 1920x1080)...")
for seg in script_segments:
    img = Image.new('RGB', (1920, 1080), color=(24, 43, 73))
    draw = ImageDraw.Draw(img)
    draw.text((100, 100), f"{seg['title'].upper()} ({seg['time']})", fill=(0, 204, 153))
    draw.text((100, 450), seg['text'], fill=(255, 255, 255))
    img.save(f"slides/{seg['id']}.png")
    print(f" Saved slide design: slides/{seg['id']}.png")

print("\nAll assets generated successfully in 'narration' and 'slides' folders.")
```

```
from moviepy.editor import ImageClip, AudioFileClip, concatenate_videoclips

# Map each segment ID to its duration in seconds based on the timeline script
timings = {
    "01_opening": 12.0,
    "02_leadership": 20.0,
    "03_communication": 20.0,
    "04_decision_making": 20.0,
    "05_problem_solving": 20.0,
    "06_business_ethics": 20.0,
    "07_closing": 8.0
}

clips = []

print("Assembling video segments...")
for seg_id, duration in timings.items():
    image_path = f"slides/{seg_id}.png"
    audio_path = f"narration/{seg_id}.mp3"

    if os.path.exists(image_path) and os.path.exists(audio_path):
        # Load image and audio, setting the explicit duration of the segment
        audio_clip = AudioFileClip(audio_path)
        # We set the image slide duration to match the audio narration duration precisely
        img_clip = ImageClip(image_path).set_duration(audio_clip.duration)
        # Bind the audio track to the static image clip
        video_segment = img_clip.set_audio(audio_clip)
        clips.append(video_segment)
        print(f" Created segment: {seg_id} (Duration: {audio_clip.duration:.2f}s)")
    else:
        print(f" Warning: Missing assets for {seg_id}")

if clips:
    # Concatenate all matching clips into a single video stream
    final_video = concatenate_videoclips(clips, method="compose")
    output_filename = "business_leadership_presentation.mp4"

    print("\nRendering video file (this may take a minute)...")
    final_video.write_videofile(
        output_filename,
        fps=24,
        codec="libx264",
        audio_codec="aac",
        temp_audiofile="temp-audio.m4a",
        remove_temp=True
    )
    print(f"\nSuccess! Video saved as: {output_filename}")
else:
    print("Error: No valid clips found to merge.")
```
